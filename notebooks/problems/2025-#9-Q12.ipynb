### ① 데이터불러오기

```python
from pathlib import Path
import pandas as pd

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)

DATASET_DIR = Path("../../assets/sources/datasets/exam_sample_data")

df = pd.read_csv(DATASET_DIR / "분개장_1급_09회_1.csv")
df.head()
```

### ② 차변 대변 금액 나누어 표시

```python
import numpy as np

df["차변금액"] = np.where(df["차대변지시자"].eq("S"), df["현지통화금액"], 0)
df["대변금액"] = np.where(df["차대변지시자"].eq("H"), df["현지통화금액"], 0)

df.head()

```

### ③ 월 데이터 생성

```python
df["월"] = pd.to_datetime(df["전표일자"], format="%Y%m%d").dt.month

df.head()
```

### ④ 계정코드 필터링

```python
df_filtered = df[df["계정코드"].eq(40001)]

df_filtered.head()
```

### ⑤ 월로 groupby 하여 정렬

```python
summary = (
    df_filtered.groupby(["월"])
      .agg(
          차변금액합계=("차변금액", "sum"),
          대변금액합계=("대변금액", "sum"),
      )
      .reset_index()
)

display(summary.sort_values(by="대변금액합계", ascending=False))
```
