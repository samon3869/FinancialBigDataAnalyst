### ① 데이터불러오기

```python
from pathlib import Path
import pandas as pd

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)

DATASET_DIR = Path("../../assets/sources/datasets/exam_sample_data")

df = pd.read_csv(DATASET_DIR / "분개장_1급_09회_1.csv")
df.head()
```

### ② 차변 대변 금액 나누어 표시

```python
import numpy as np

df["차변금액"] = np.where(df["차대변지시자"].eq("S"), df["현지통화금액"], 0)
df["대변금액"] = np.where(df["차대변지시자"].eq("H"), df["현지통화금액"], 0)

df.head()

```

### ③ 라인별 정상매출 요건 칼럼 정의

```python
REVENUE_ACCT_CODES = {40001, 40002, 40004, 40005, 40007, 40009, 40011, 40013}
REVENUE_COUNTERPART_ACCT_CODES = {10003,10004,10005,10006,10009,10011,10012, 10013,10015,10017,10018,10021,10022,10023,10026,10027,10028,10032,10036,10037,10038,10039,10040,10044,10048,10050,10053,10054,10055,10057,10059,10060,10061,10063,10064,10065,10066,10067,10080,10084,10086,10087,10092,20017}

is_revenue_normal = (
    df["계정코드"].isin(REVENUE_ACCT_CODES)
    & ((df["차변금액"] < 0) | (df["대변금액"] > 0))
    )

is_counterpart_normal = (
    df["계정코드"].isin(REVENUE_COUNTERPART_ACCT_CODES)
    & ((df["차변금액"] > 0) | (df["대변금액"] < 0))
    )


df["REVENUE_FLAG"] = is_revenue_normal.astype("int")
df["COUNTERPART_FLAG"] = is_counterpart_normal.astype("int")


df.head()
```

### ④ 전표세트PK로 groupby 후 전표세트별 요건판단

```python
voucher_summary = (
    df.groupby(["전표일자", "전표번호"])
        .agg(
            REVENUE_FLAG_SUM=("REVENUE_FLAG", "sum"),
            COUNTERPART_FLAG_SUM=("COUNTERPART_FLAG", "sum"),
        )
        .reset_index()
)

voucher_summary.head()
```

### ⑤ 비정상거래 출력

```python
abnormal_vouchers = voucher_summary[
    (voucher_summary["REVENUE_FLAG_SUM"] > 0) &
    (voucher_summary["COUNTERPART_FLAG_SUM"] == 0)
]

len(abnormal_vouchers)
```
